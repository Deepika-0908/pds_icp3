```python
import pandas as pd

# Load the Sacramento housing dataset from the provided URL
url = "https://raw.githubusercontent.com/JuliaData/CSV.jl/refs/heads/main/test/testfiles/Sacramentorealestatetransactions.csv"
df = pd.read_csv(url)

# Normalize column names (lowercase and strip whitespace)
df.columns = df.columns.str.strip().str.lower()

# Display the column names
print("Column names in dataset:")
print(df.columns)

# Display the first 5 rows of the dataset
print("\nFirst 5 rows of the dataset:")
print(df.head())

# 1. Number of columns
print("\nNumber of columns:", df.shape[1])

# 2. Mean house price
print("Mean house price:", df['price'].mean())

# 3. Median house price
print("Median house price:", df['price'].median())

# 4. Different types of property
print("Different types of property:", df['type'].unique())

# 5. Maximum price of a Condo
max_condo_price = df[df['type'].str.lower() == 'condo']['price'].max()
print("Maximum price of a Condo:", max_condo_price)

# 6. Mean price for each type of property
mean_price_by_type = df.groupby('type')['price'].mean()
print("\nMean price by property type:")
print(mean_price_by_type)

# 7. Count of different property types
property_type_counts = df['type'].value_counts()
print("\nCount of each property type:")
print(property_type_counts)
```
